```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
trial_data_vglut1_file = "/Volumes/Intenso/image-analysis/synapse-counting/VCAM1/VCAM1-LacZ_VGLUT1-PSD95_output_data/local_peak_optimization_trial_data.csv"
trial_data_vglut1 = pd.read_csv(trial_data_vglut1_file)

trial_data_vglut1_core = trial_data_vglut1[["hippocampal_layer", "value", "number"]]
trial_data_vglut1_core
```

```python
# pivoting
trial_data_vglut1_wide = trial_data_vglut1_core.pivot(index = "number", columns="hippocampal_layer", values="value")
trial_data_vglut1_wide
```

```python
palette = sns.color_palette("pastel", n_colors=trial_data_vglut1_wide.shape[1])

fig, ax = plt.subplots(figsize=(10, 8)) 
lines = trial_data_vglut1_wide.plot.line(color = palette, ax = ax)
ax.set_title("VGLUT1-PSD95, local peaks maxima, parameter optimization")
ax.set_xlabel("Trial number", fontsize=10)
ax.set_ylabel("Mean scaled difference", fontsize=10)

ax.legend(
    title="Hippocampal layer",
    loc='center left',
    bbox_to_anchor=(1.02, 0.5),  
    borderaxespad=0,
    fontsize=9,
    title_fontsize=10
)

plt.tight_layout()
plt.show()
```
