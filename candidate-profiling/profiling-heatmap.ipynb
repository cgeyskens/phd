```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm
```

```python
df = pd.DataFrame(
    {
        "Spatial transcriptomics":  ["2","2","2","2","2","2","2","2","2","2"],
        "Working-Ab-for-IHC-WB":    ["2","2","2","1","0","2","2","2","2","2"],
        "IHC-aldh1l1-eGFP":         ["2","2","2","0","0","1","2","1","1","2"],
        "IHC-aldh1l1-GLT1":         ["2","2","2","0","0","1","2","1","1","2"],
        "smFISH":                   ["2","2","2","2","0","0","2","0","0","2"],
        "Synaptogliosomes":         ["2","2","2","0","0","0","2","0","0","2"],
        "WB-dev-analysis":          ["2","2","2","0","0","0","0","0","0","0"],
    },
    index=[
        "Gpr37l1",
        "Hepacam",
        "Vcam1",
        "Plxnb1",
        "Ephb4",
        "Slc3a2",
        "Cadm4",
        "Neo1",
        "Lsamp",
        "Alcam"
    ]
)
```

```python
df = df.astype(int) 
cmap = ListedColormap([
    "#F0F0F0", # not tested
    "#F4A3A3",  # tested & absent
    "#A6D9D0" # tested & present
    ]) 
norm = BoundaryNorm([-0.5, 0.5, 1.5, 2.5], cmap.N)
```

```python
plt.figure(figsize=(8, 4))
ax = sns.heatmap(
    df,
    cmap=cmap,
    norm=norm,
    fmt="",
    linewidths=0.5,
    linecolor="white",
    cbar=True
)
cbar = ax.collections[0].colorbar
cbar.set_ticks([0, 1, 2])
cbar.set_ticklabels(["not tested", "tested, no/absent", "tested, yes/present"])

# plt.xlabel("")
# plt.ylabel("")
plt.tight_layout()

plt.savefig(
    "heatmap_candidates.svg",
    format="svg",
    bbox_inches="tight"
)
plt.show()
```
